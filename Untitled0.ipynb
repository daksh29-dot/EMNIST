```python
import tensorflow as tf
import tensorflow_datasets as tfds

# Load EMNIST 'digits' split
# The first time you run this, it will download the dataset, which might take a few minutes.
(ds_train, ds_test), ds_info = tfds.load(
    'emnist/digits',
    split=['train', 'test'],
    shuffle_files=True,
    as_supervised=True,
    with_info=True
)

num_examples_to_use = 300000 # 3 lakh examples

# Limit the number of examples in the training and test datasets
ds_train = ds_train.take(num_examples_to_use)
ds_test = ds_test.take(num_examples_to_use)

def preprocess_emnist(image, label):
    # Convert image to float32 and normalize to [0, 1]
    image = tf.cast(image, tf.float32) / 255.0
    # EMNIST images are already provided in (28, 28, 1) format by tfds for 'digits' split.
    # No transposition or additional channel dimension expansion is needed.
    return image, label

# Apply preprocessing and batching
ds_train = ds_train.map(preprocess_emnist).batch(32).prefetch(tf.data.AUTOTUNE)
ds_test = ds_test.map(preprocess_emnist).batch(32).prefetch(tf.data.AUTOTUNE)

print("EMNIST dataset loaded and preprocessed.")
print(f"Number of classes in EMNIST: {ds_info.features['label'].num_classes}")
print(f"Using {num_examples_to_use} examples for training and testing each (if available in the dataset).")

# Displaying the shape of one batch of images
for image_batch, label_batch in ds_train.take(1):
    print(f"Shape of one image batch: {image_batch.shape}")
    print(f"Shape of one label batch: {label_batch.shape}")
```

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense, Dropout

# Define the CNN model
model = Sequential([
    Conv2D(32, (3, 3), activation='relu', input_shape=(28, 28, 1)),
    MaxPooling2D((2, 2)),
    Conv2D(64, (3, 3), activation='relu'),
    MaxPooling2D((2, 2)),
    Flatten(),
    Dense(128, activation='relu'),
    Dropout(0.5),
    Dense(10, activation='softmax')  # 10 classes for digits 0-9
])

# Compile the model
model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])

print("CNN model defined and compiled successfully.")
```

### Train the CNN Model

Now that the model is defined, we need to train it using the preprocessed EMNIST dataset. This will teach the model to recognize digits.

```python
# Train the model
# You can adjust the number of epochs and add callbacks as needed.
# For a quick demonstration, we'll start with a few epochs.
print("Training the model...")
history = model.fit(
    ds_train,
    epochs=5, # You can increase this for better accuracy
    validation_data=ds_test
)

print("Model training complete.")
```

```python
# Use tf.data.experimental.cardinality to get the number of elements in the dataset
# Note: This will show the number of *batches* if the dataset has been batched.
# To get the number of examples, we need to consider the batch size.

# If the dataset is already batched, cardinality gives the number of batches.
# For example, if you have 300,000 examples and a batch size of 32:
# 300,000 / 32 = 9375 batches

# To get the number of examples from a batched dataset, you can do:
# num_batches * batch_size, or get the cardinality before batching.

# Assuming the batch size is 32 (as defined in the previous cell)
batch_size = 32

# For the training dataset
try:
    num_train_batches = tf.data.experimental.cardinality(ds_train).numpy()
    # Check if cardinality is unknown (e.g., if prefetch is applied before take)
    if num_train_batches == tf.data.experimental.INFINITE_CARDINALITY:
        print("Training dataset cardinality is infinite. Cannot determine exact number of examples easily after prefetch.")
        print("However, it is limited to 300,000 examples by the .take() operation before batching and prefetching.")
    else:
        num_train_examples = num_train_batches * batch_size
        print(f"Effective number of examples in ds_train (after .take(300000)): {num_train_examples}")
except Exception as e:
    print(f"Could not determine training dataset cardinality: {e}")
    print("It is still limited to 300,000 examples by the .take() operation.")

# For the test dataset
try:
    num_test_batches = tf.data.experimental.cardinality(ds_test).numpy()
    if num_test_batches == tf.data.experimental.INFINITE_CARDINALITY:
        print("Test dataset cardinality is infinite. Cannot determine exact number of examples easily after prefetch.")
        print("However, it is limited to 300,000 examples (or fewer if original test set is smaller) by the .take() operation before batching and prefetching.")
    else:
        num_test_examples = num_test_batches * batch_size
        print(f"Effective number of examples in ds_test (after .take(300000)): {num_test_examples}")
except Exception as e:
    print(f"Could not determine test dataset cardinality: {e}")
    print("It is still limited to 300,000 examples (or fewer if original test set is smaller) by the .take() operation.")
```

```python
# Import necessary libraries for image processing
from google.colab import files
from PIL import Image
import numpy as np
import matplotlib.pyplot as plt
import io
```

```python
# Upload a file from your local system
uploaded = files.upload()

# Assuming only one file is uploaded
for filename in uploaded.keys():
    custom_image_filename = filename # Correctly assign the uploaded filename
    print(f'Uploaded file: {custom_image_filename}')
```

### Load and Preprocess the Custom Image

Now, let's load the uploaded image, preprocess it to match the model's input requirements, and display it.

```python
try:
    # Load the image
    img = Image.open(io.BytesIO(uploaded[custom_image_filename]))

    # Convert to grayscale
    img = img.convert('L')

    # Resize to 28x28 pixels
    img = img.resize((28, 28))

    # Invert colors if necessary (MNIST has white digits on black background)
    # This step might need adjustment based on your image. If your digit is black on white,
    # uncomment the line below:
    # img = Image.eval(img, lambda x: 255 - x)

    # Convert to numpy array
    img_array = np.array(img)

    # Normalize pixel values to 0-1
    img_array = img_array / 255.0

    # Reshape to add batch and channel dimensions (1, 28, 28, 1)
    custom_image_processed = img_array.reshape(1, 28, 28, 1)

    # Display the processed image
    plt.imshow(custom_image_processed.reshape(28, 28), cmap='gray')
    plt.title('Processed Custom Image')
    plt.axis('off')
    plt.show()

    print("Custom image processed successfully!")
except Exception as e:
    print(f"Error processing custom image: {e}")
    print("Please ensure you have uploaded a valid image file.")
```

### Make a Prediction

Now, let's use our trained model to predict the digit in your custom image.

```python
if 'custom_image_processed' in locals():
    # Make prediction
    custom_prediction = model.predict(custom_image_processed)

    # Get the predicted digit
    predicted_digit = custom_prediction.argmax()

    print(f"The model predicts the digit is: {predicted_digit}")
    print(f"Prediction probabilities: {custom_prediction}")
else:
    print("No custom image was successfully processed to make a prediction.")
```
